```python
import os, sys
from bertviz import head_view, model_view
from transformers import RobertaTokenizer
from transformers import RobertaForMaskedLM, RobertaForSequenceClassification


sys.path.append("../")
from utils.parameters import *
```

```python
tokenizer = RobertaTokenizer.from_pretrained(TOKEN_DIR_BPE)
# tokenizer = BertTokenizer.from_pretrained(TOKEN_DIR_BPE)


sentence_a = "CCCTAAACCCTAAACCCTAACCCTAACCCTAACCCTAACCC"
sentence_b = "CCCTCAGCCGGCCCGCCCGCCC"
print(tokenizer.tokenize(sentence_a))
print(tokenizer.convert_tokens_to_ids(tokenizer.tokenize(sentence_a)))
print(tokenizer.tokenize(sentence_b))
print(tokenizer.convert_tokens_to_ids(tokenizer.tokenize(sentence_b)))
sentence_c = "CCCTAACCCTAACCCTAACCCTAACCCTAACCCTAACCCC"
print(tokenizer.tokenize(sentence_c))
print(tokenizer.convert_tokens_to_ids(tokenizer.tokenize(sentence_c)))

inputs = tokenizer.encode_plus(sentence_c, return_tensors='pt')
print(inputs)
```

```python
# 4 layers, 8 heads
# model = BertModel.from_pretrained(output_dir)
pretrained_model1 = RobertaForMaskedLM.from_pretrained(PRETRAINED_MODEL_DIR_BPE_4L8HT2)

# Get the model's attention weights
outputs1 = pretrained_model1(**inputs, output_attentions=True)
attention1 = outputs1.attentions

# Use BertViz to visualize the attention
tokens1 = tokenizer.convert_ids_to_tokens(inputs['input_ids'][0])
print(tokens1)
head_view(attention1, tokens1)
```

```python
# Use BertViz to visualize the attention with model_view
model_view(attention1, tokens1)
```

```python
# 4 layers, 8 heads
finetuned_model1 = RobertaForSequenceClassification.from_pretrained(FINETUNING_MODEL_DIR_BPE_4L8HT2)


# Get the model's attention weights
outputs2 = finetuned_model1(**inputs, output_attentions=True)
print(outputs2)
attention2 = outputs2.attentions

# Use BertViz to visualize the attention
tokens2 = tokenizer.convert_ids_to_tokens(inputs['input_ids'][0])
print(tokens2)
head_view(attention2, tokens2)
```

```python
# Use BertViz to visualize the attention with model_view
model_view(attention2, tokens2)
```
